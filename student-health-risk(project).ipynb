# Feature Engineering

```python
bmi_deviation = (np.maximum(0,18.5 - df["bmi"])+np.maximum(0,df["bmi"] - 24.9))

df["bmi_risk"] = np.clip(bmi_deviation / 10,0,1)

df["stress_risk"] = (df["stress_level"].map({"low": 0.0,"medium": 0.5,"high": 1.0}))

df["sleep_duration_risk"] = np.clip(abs(df["sleep_duration"] - 8) / 4,0,1)

step_risk = (1 -np.clip(df["step_count"] / 10000,0,1))

exercise_risk = (1 -np.clip(df["exercise_duration"] / 60,0,1))

calorie_risk = (1 -np.clip(df["calorie_expenditure"] / 600,0,1))

df["activity_risk"] = (0.40 * step_risk+0.30 * exercise_risk+0.30 * calorie_risk)

heart_rate = df["heart_rate"]

df["heart_rate_risk"] = np.clip(np.where(heart_rate < 60,(60 - heart_rate) / 20,np.where(heart_rate > 100,(heart_rate - 100) / 40,0)),0,1)

df["water_risk"] = (1 -np.clip(df["water_intake"] / 3,0,1))

df["sleep_quality_risk"] = (df["sleep_quality"].map({"poor": 1.0,"average": 0.5,"good": 0.0}))

df["lifestyle_risk"] = (df["smoking_alcohol"].map({"no": 0.0,"occasional": 0.5,"yes": 1.0}))
```

```python

risk_columns = ["bmi_risk","stress_risk","sleep_duration_risk","activity_risk","heart_rate_risk","water_risk","sleep_quality_risk","lifestyle_risk"]

print("nan values in risk features:")

print(df[risk_columns].isnull().sum())
```

```python

df["health_risk_score"] = (100 *(0.20 * df["bmi_risk"]+0.15 * df["stress_risk"]+0.15 * df["sleep_duration_risk"]+0.20 * df["activity_risk"]+

    0.10 * df["heart_rate_risk"]+0.10 * df["water_risk"]+0.05 * df["sleep_quality_risk"]+0.05 * df["lifestyle_risk"]))
```

```python
df = df.dropna(subset=["health_condition","health_risk_score"]).reset_index(drop=True)
```

```python
df["health_risk_score"] = np.clip(df["health_risk_score"],0,100)
```

# Train_Test_Split

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_class_train, y_class_test, y_reg_train, y_reg_test = train_test_split(X,y_class,y_reg,test_size=0.20,random_state=23,stratify=y_class)
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
label_encoder = LabelEncoder()
y_class_train_enc = (label_encoder.fit_transform(y_class_train))
y_class_test_enc = (label_encoder.transform(y_class_test))
```

# splitting the data based on datatype for regression

```python
num_cols = ["sleep_duration","heart_rate","bmi","calorie_expenditure","step_count","exercise_duration","water_intake","bmi_risk","stress_risk",
            "sleep_duration_risk","activity_risk","heart_rate_risk","water_risk","sleep_quality_risk","lifestyle_risk"]
```

```python
nominal_cols = ["diet_type","smoking_alcohol","gender"]
```

```python
ord_cols = ["stress_level","sleep_quality","physical_activity_level"]
```

```python
ordinal_categories = [["low","medium","high"],["poor","average","good"],["sedentary","moderate","active"]]
```

# Data Preprocessing

```python
from sklearn.impute import SimpleImputer
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
from sklearn.preprocessing import OneHotEncoder,OrdinalEncoder
```

```python
from sklearn.pipeline import Pipeline
```

```python
from sklearn.compose import ColumnTransformer
```

```python
num_pipeline = Pipeline([("imputer",SimpleImputer(strategy="median")),
                         ("scaler",StandardScaler())])
```

```python
nominal_pipeline = Pipeline([("imputer",SimpleImputer(strategy="most_frequent")),
                             ("encoder",OneHotEncoder(handle_unknown="ignore",drop="first"))])
```

```python
ordinal_pipeline = Pipeline([("imputer",SimpleImputer(strategy="most_frequent")),
                           ("encoder",OrdinalEncoder(categories=ordinal_categories))])
```

```python
preprocessing = ColumnTransformer([("numerical",num_pipeline,num_cols),
                                   ("nominal",nominal_pipeline,nominal_cols),
                                   ("ordinal",ordinal_pipeline,ord_cols)])
```

```python
X_train_p = (preprocessing.fit_transform(X_train))

X_test_p = (preprocessing.transform(X_test))

y_class_train_enc = np.asarray(y_class_train_enc, dtype="int32")

y_class_test_enc = np.asarray(y_class_test_enc,dtype="int32")

y_reg_train = np.asarray(y_reg_train,dtype="float32")

y_reg_test = np.asarray(y_reg_test,dtype="float32")
```

# Architecture Building

```python
from keras.layers import Input,Dense,Dropout

from tensorflow.keras.models import Model

from tensorflow.keras.optimizers import Adam

from tensorflow.keras.callbacks import EarlyStopping

from sklearn.utils.class_weight import compute_class_weight
```

```python
input_layer = Input(shape=(X_train_p.shape[1],))

h1 = Dense(96,activation="relu")(input_layer)
h1 = Dropout(0.30)(h1)

h2 = Dense(32,activation="relu")(h1)
h2 = Dropout(0.20)(h1)

o= Dense(16,activation="relu")(h2)


classification_output = Dense(3,activation="softmax",name="classification")(o)


regression_output = Dense(1,activation="linear",name="regression")(o)


modelf = Model(inputs=input_layer,outputs=[classification_output,regression_output])
modelf.summary()
```

```python
modelf.compile(optimizer=Adam(learning_rate=0.001),loss=["sparse_categorical_crossentropy","mse"],metrics=[["accuracy"],["mae"]])
```

```python
weights= compute_class_weight(class_weight="balanced",classes=np.unique(y_class_train_enc),y=y_class_train_enc)
class_weight_dict = dict(zip(np.unique(y_class_train_enc),weights))
```

```python
classification_sample_weight = np.array([class_weight_dict[int(value)]
                for value in y_class_train_enc],dtype="float32")

regression_sample_weight = np.ones(len(y_reg_train),dtype="float32")
```

```python
early_stop = EarlyStopping(monitor="val_loss",patience=5,min_delta=0.0001,restore_best_weights=True)
```

```python
history = modelf.fit(X_train_p,[y_class_train_enc,y_reg_train],validation_data=(X_test_p,[y_class_test_enc,y_reg_test]),
     sample_weight=[classification_sample_weight,regression_sample_weight],epochs=10,batch_size=256,callbacks=[early_stop],verbose=1)
```

# Prediction

```python
predictions = modelf.predict(X_test_p)

classification_predictions = (predictions[0])

regression_predictions = (predictions[1])
```

```python
predicted_class_index = np.argmax(classification_predictions,axis=1)

predicted_class = (label_encoder.inverse_transform(predicted_class_index))
print(predicted_class)
print(regression_predictions)
```

# Evaluation
